```python
import pandas as pd

#import csv as dataframes
people_data_df = pd.read_csv("../data/processed/people_data_cleaned.csv")
evasive_df = pd.read_csv("../data/processed/evasive.csv")
events_df = pd.read_csv("../data/processed/events.csv")
processed_statements_df = pd.read_csv("../data/processed/processed_statements.csv")
score_table_df = pd.read_csv("../data/processed/score_table.csv")
security_log_df = pd.read_csv("../data/processed/security_logs_cleaned.csv")

#compare student details in testimony to people data
for x in (processed_statements_df, people_data_df, score_table_df):
    x["Name"] = x["Name"].str.lower()

for x in (processed_statements_df, people_data_df, score_table_df):
    x["Student_ID"] = x["Student_ID"].str.upper()

merged = processed_statements_df.merge(people_data_df, on="Name", how="left", suffixes=("_1", "_2"), indicator=True)

mismatch = (merged["_merge"] == "left_only") | (merged["Student_ID_1"] != merged["Student_ID_2"])
mismatched_names = merged.loc[mismatch, "Name"]

score_table_df.loc[score_table_df["Name"].isin(mismatched_names), "Score"] += 1

#flag evasive behaviour
evasive_people = evasive_df.loc[evasive_df["Evasive"], "Name"]
score_table_df.loc[score_table_df["Name"].isin(evasive_people), "Score"] += 1


#compare student location with security logs

#compare student meetings

#compare student location with location data


#save dataframes to csv
score_table_df.to_csv("../data/processed/score_table.csv", index=False)
```
