```python
import pandas as pd

#people_data.csv
people_data_df = pd.read_csv("../data/raw/people_data.csv")
people_data_df = people_data_df.rename(columns={"Student ID": "Student_ID", 
                                                "Year of Study": "Year_of_Study", 
                                                "Hair colour": "Hair_Colour"})

#location_data.csv
location_data_df = pd.read_csv("../data/raw/location_data.csv")
location_data_df = location_data_df.rename(columns={"Building Name": "Building_Name", 
                                                "Opening Times": "Opening_Times"})

#security_logs.csv
security_logs_df = pd.read_csv("../data/raw/security_logs.csv")
security_logs_df = security_logs_df.rename(columns={"Student ID": "Student_ID"})
security_logs_df[["Start_Time", "End_Time"]] = security_logs_df["Time"].str.split("-", expand=True)
security_logs_df = security_logs_df.drop(columns="Time")

people_data_df.to_csv("../data/processed/people_data_cleaned.csv", index=False)
location_data_df.to_csv("../data/processed/location_data_cleaned.csv", index=False)
security_logs_df.to_csv("../data/processed/security_logs_cleaned.csv", index=False)
```
