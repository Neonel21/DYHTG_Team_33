```python
import dateparser
import pandas as pd
import json

with open("../data/processed/processed_testimonies.json", "r", encoding="utf-8") as f:
    data = json.load(f)

events_df = pd.DataFrame(
    [
        {
            "Student_ID": info["id"],
            "Name": name,
            "Location": event["place"],
            "Start_Time": event["start"],
            "End_Time": event["end"],
        }
        for name, info in data.items()
        for event in info["events"]
    ]
)

evasive_df = pd.DataFrame(
    [
        {"Student_ID": info["id"], "Name": name, "Evasive": info["evasive"]}
        for name, info in data.items()
    ]
)


events_df.to_csv("../data/processed/events.csv", index=False)
evasive_df.to_csv("../data/processed/evasive.csv", index=False)
```
