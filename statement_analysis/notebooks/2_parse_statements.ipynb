```python
import re
import pandas as pd

#create scoring table
people_data_df = pd.read_csv("../data/processed/people_data_cleaned.csv")
score_table_df = people_data_df[["Student_ID", "Name"]].copy()
score_table_df["Score"] = 0

#import statements file
file_path = "../data/raw/student_statements.txt"

with open(file_path, "r", encoding="utf-8") as f:
    raw_statements = f.read()

#split by statements
pattern = r"^[ \t]*Statement[ \t]+\d+[ \t]*$"
blocks = re.split(pattern, raw_statements, flags=re.MULTILINE)

statements = [block.strip() for block in blocks if block.strip()]
    
#extract student numbers
pattern = r"^Student Number:\s*(.*)$"
student_ids = []
for statement in statements:
    student_ids.append(re.search(pattern, statement, re.MULTILINE).group(1))

#extract student names
pattern = r"^(?:Name|Student Name):\s*(.*)$"
student_names = []
for statement in statements:
    student_names.append(re.search(pattern, statement, re.MULTILINE).group(1))

#extract testimonies
pattern = r"^Testimony:\s*(.*)"
testimonies = []
for statement in statements:
    testimonies.append(re.search(pattern, statement, re.MULTILINE).group(1))

#assemble data into dataframe and save as csv
processed_statements_df = pd.DataFrame({
    "Student_ID": student_ids,
    "Name": student_names,
    "Testimony": testimonies,
})

score_table_df.to_csv("../data/processed/score_table.csv", index=False)
processed_statements_df.to_csv("../data/processed/processed_statements.csv", index=False)
```
